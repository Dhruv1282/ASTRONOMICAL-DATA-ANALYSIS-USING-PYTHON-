#Control functions (loops)

```python
# list will be made by []

# loops

# for loops

for i in [0,1,2,3,4,5,6]:
  print(i)


```

```python
# we can also use range(start,end,skip) so we dont need to make list here
#here start is includede but the end is not included

for i in range(0,10):
  print(i)

print("For negative numbers also:\n")

for i in range(-5,10):
  print(i)
```

#While loop

```python
# making a input which takes favourite celestial objects and update the index
i=1
while True:
  object=input("Enter your favourite celestial object: ")
  print(f"{i}. {object}")
  i=i+1

  if object.lower() =="end":    #lower for converting the input to lower case
    break
```
